````python
# TODO: متدی برای پیش پردازش  فضای ذخیره سازی طبق توضیحات ارائه شده بنویسید.

def parse_storage(storage_str):
    capacity = storage_str.split(' ')[0]
    if 'TB' in capacity:
        capacity_gb = float(capacity.replace('TB', '')) * 1024
    elif 'GB' in capacity:
        capacity_gb = float(capacity.replace('GB', ''))
    else:
        capacity_gb = 0    
    
    if 'SSD' in storage_str:
        storage_type = 'SSD'
    elif 'HDD' in storage_str:
        storage_type = 'HDD'
    else:
        storage_type = 'Other'

    return capacity_gb, storage_type

df_processed[['Storage_Capacity_GB', 'Storage_Type']] = df_processed['Storage'].apply(parse_storage).apply(pd.Series)
df_processed = df_processed.drop('Storage', axis=1)
df_processed.head()
````

````python
# TODO متدی برای مدیریت ویژگی  Processor بنویسید.

# مهمترین عامل پروسسور علاوه بر برند پروسسور نسل پروسسور است مثلا نسل ۳ یا نسل ۷ که از برند پروسسور مهمتر است

import re
def parse_processor(processor_str):
    if not isinstance(processor_str, str):
        return 'Unknown', 0
    parts = processor_str.split()
    brand = parts[0]
    series = 0
    match = re.search(r'(i[3579])|(Ryzen\s+[3579])', processor_str, re.IGNORECASE)
    if match:
        series_str = match.group(0)
        num_match = re.search(r'\d+', series_str)
        if num_match:
            series = int(num_match.group(0))
    return brand, series

df_processed[['Processor_Brand', 'Processor_Series']] = df_processed['Processor'].apply(lambda x: pd.Series(parse_processor(x)))
df_processed = df_processed.drop('Processor', axis=1)
df_processed.head()
````

````python
# TODO متدی برای مدیریت ویژگی GPU بنویسید.

# برای گارت گرافی ۳ مدل داریم انویدیا ای ام دی و مدار مجتمع

def parse_gpu(gpu_str):
    if not isinstance(gpu_str, str):
        return 'Unknown', 'Unknown'
    parts = gpu_str.split()
    brand = parts[0]
    if len(parts) == 1:
        if brand.lower() == 'integrated':
            return 'Integrated', 'Integrated'
        else:
            return brand, 'Unknown'
    else:
        if brand.lower() in ['nvidia', 'amd']:
            series = " ".join(parts[1:])
            if 'GTX' in series or 'RTX' in series or 'RX' in series:
                return brand, parts[1]
            else:
                return brand, 'Other_Dedicated' 
        elif brand.lower() == 'integrated':
            return 'Integrated', 'Integrated'
        else:
            return brand, " ".join(parts[1:])


df_processed[['GPU_Brand', 'GPU_Series']] = df_processed['GPU'].apply(lambda x: pd.Series(parse_gpu(x)))
df_processed = df_processed.drop('GPU', axis=1)
df_processed.head()
````

````python
# TODO: متد منحنی یادگیری را بنویسید

def plot_learning_curve(model, X, y, model_name):
    train_sizes, train_scores, val_scores = learning_curve(model, X, y, cv=5, scoring='neg_mean_squared_error')
    train_mean = -np.mean(train_scores, axis=1)
    val_mean = -np.mean(val_scores, axis=1)
    plt.plot(train_sizes, train_mean, label='Training Error')
    plt.plot(train_sizes, val_mean, label='Validation Error')
    plt.title(f'Learning Curve for {model_name}')
    plt.xlabel('Training Size')
    plt.ylabel('MSE')
    plt.legend()
    plt.show()
````

````python
# TODO: تابع برای محاسبه و نمایش معیارهای ارزیابی

def evaluate_model(pipeline, X, y, dataset_name):
    y_pred = pipeline.predict(X)
    mse = mean_squared_error(y, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y, y_pred)
    r2 = r2_score(y, y_pred)
    print(f"Model Evaluation On Sets: {dataset_name}:")
    print(f"Mean Squared Error (MSE): {mse:.4f}")
    print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
    print(f"Mean Absolute Error (MAE): {mae:.4f}")
    print(f"R-squared (R²): {r2:.4f}")
    return mse, rmse, mae, r2
````

````python
# TODO: نمایش نمودار پراکندگی بین مقادیر واقعی و پیش‌بینی شده

def plot_prediction_scatter(pipeline, X_train, y_train, X_val, y_val, X_test, y_test):
    fig, axs = plt.subplots(1, 4, figsize=(20, 5))
    for ax, X, y, title in zip(axs, [X_train, X_val, X_test], [y_train, y_val, y_test], ['Train', 'Validation', 'Test']):
        y_pred = pipeline.predict(X)
        ax.scatter(y, y_pred, alpha=0.5)
        ax.plot([y.min(), y.max()], [y.min(), y.max()], 'r--')
        ax.set_title(f'{title} Predictions')
        ax.set_xlabel('Actual')
        ax.set_ylabel('Predicted')
    axs[3].hist(y_test - pipeline.predict(X_test), bins=30)
    axs[3].set_title('Test Error Distribution')
    plt.tight_layout()
    plt.show()
````
